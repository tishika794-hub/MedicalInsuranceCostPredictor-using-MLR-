# Medical Insurance Data Cleaning
This notebook loads the `insurance.csv` dataset, checks for missing values, and encodes categorical variables.

```python
import pandas as pd
import numpy as np

# Load the data
df = pd.read_csv('insurance.csv')
df.head()
```

```python
# Check for missing values
print("Missing values in each column:\n", df.isnull().sum())
```

```python
# Check data types
df.info()
```

```python
# Encode categorical variables
df['sex'] = df['sex'].map({'male': 0, 'female': 1})
df['smoker'] = df['smoker'].map({'yes': 1, 'no': 0})

# One-hot encode the 'region' column
df = pd.get_dummies(df, columns=['region'], drop_first=True)

df.head()
```

```python
# Save the cleaned dataset
df.to_csv('insurance_cleaned.csv', index=False)
print("Cleaned dataset saved as 'insurance_cleaned.csv'")
```
